```python
import matplotlib.pyplot as plt
import numpy as np

def calculate_orbits(n1, n2):
    mod = n1 * n2 - 1
    orbits = [[] for _ in range(mod)]
    dones = set()
    
    for i in range(1, mod):
        if i not in dones:
            orb = [i]
            cur = (i * n2) % mod
            while cur != i:
                orb.append(cur)
                cur = (cur * n2) % mod
            for o in orb:
                dones.add(o)
                orbits[i].append((np.cos(o * (2 * np.pi / mod)), np.sin(o * (2 * np.pi / mod))))
    non_empty_lists = [lst for lst in orbits if lst]
    return non_empty_lists

def plot_orbits(ax, orbits, width):
    for orbit in orbits:
        if orbit:
            x_coords, y_coords = zip(*orbit)
            ax.plot(x_coords + (x_coords[0],), y_coords + (y_coords[0],), lw=width)

def Picture_2v2(n1, n2, size, width, res=100, style='seaborn-darkgrid', back="black"):
    fig, ax = plt.subplots(figsize=(size, size), dpi=res)
    plt.style.use(style)
    ax.set_facecolor(back)
    
    orbits = calculate_orbits(n1, n2)
    plot_orbits(ax, orbits, width)
    numb=len(orbits)
    
    plt.xlim([-1.1, 1.1])
    plt.ylim([-1.1, 1.1])
    plt.yticks([])
    plt.xticks([])
    ax.set_aspect('equal')
    plt.grid(False)
    plt.title("((1,{}),(2,{})) and ((2,{}),(1,{}))    No. of Orbits: {}".format(n1, n2, n2, n1,numb))
    plt.show()
    

    
    
def calculate_orbits_Single(n1, n2, mod):
    orbits = []
    dones = set()
    for i in range(mod):
        if i not in dones:
            orb = [i]
            cur = (i * n2) % mod
            while cur != i:
                orb.append(cur)
                cur = (cur * n2) % mod
            orbits.append(orb)
            dones.update(orb)
    return orbits

def plot_orbit_Single(ax, orbit, mod, width, col):
    x_coords = [np.cos(o * (2 * np.pi / mod)) for o in orbit]
    y_coords = [np.sin(o * (2 * np.pi / mod)) for o in orbit]
    ax.plot(x_coords, y_coords, lw=width, color=col)

def Picture_2v2_Single(n1, n2, frame, size, width, col='r', res=100):
    mod = n1 * n2 - 1
    orbits = calculate_orbits_Single(n1, n2, mod)
    
    fig, ax = plt.subplots(figsize=(size, size), dpi=res)
    plt.style.use('seaborn-notebook')
    ax.set_facecolor("black")
    
    for i, orbit in enumerate(orbits):
        if i == frame:
            plot_orbit_Single(ax, orbit, mod, width, col)
            break

    plt.xlim([-1.1, 1.1])
    plt.ylim([-1.1, 1.1])
    plt.yticks([])
    plt.xticks([])
    ax.set_aspect('equal')
    plt.grid(False)
    plt.title("((1,{}),(2,{})) and ((2,{}),(1,{}))    Cyclic Orbit of {}".format(n1, n2, n2, n1, i))
    plt.show()
```
